```python
sentences = [

    ["python", "programming", "language"],
    ["machine", "learning", "artificial", "intelligence"],
    ["sql", "database", "query"],
    ["deep", "learning", "neural", "network"],
    ["data", "science", "python"]

]

print("Dataset Loaded")
```

```python
from gensim.models import Word2Vec

model = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4
)

print("✅ Word Embedding Model Trained")
```

```python
word = "python"

vector = model.wv[word]

print(f"Word: {word}")

print("\nEmbedding Vector:\n")

print(vector)
```

```python
def word_embedding_agent(word):

    try:

        print(f"\nWord: {word}")

        print("\nTop Similar Words:\n")

        similar_words = model.wv.most_similar(
            word,
            topn=5
        )

        for w, score in similar_words:

            print(
                f"{w}  --->  {round(score,3)}"
            )

    except:

        print("Word Not Found")
```

```python
while True:

    word = input(
        "\nEnter Word: "
    )

    if word.lower() == "exit":
        break

    word_embedding_agent(word)
```
